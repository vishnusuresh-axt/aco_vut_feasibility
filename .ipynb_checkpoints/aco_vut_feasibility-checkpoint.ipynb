# Tafamidis

## Step 1

```python
tafamidis_codes = list(pd.read_excel("Aco_Vutri_TriNetX_Feas._codelists_JUL2026.xlsx", sheet_name = "Tafamidis")["Concept Code"])
```

```python
tafamidis_codes = [str(code) for code in tafamidis_codes]
```

```python
# STEP 1: Get first tafamidis code on/after 2019-06-01
step1_df = (
    session.table("DATA_202511.MEDICATION")
    .filter(col("CODE_SYSTEM") == "RxNorm")
    .filter(col("CODE").isin(tafamidis_codes))
    .filter(col("START_DATE") >= "2019-06-01")
    .filter(col("START_DATE").is_not_null())
    .group_by("PATIENT_ID")
    .agg(s_min("START_DATE").alias("INDEX_DATE"))
)
print(f"Step 1: {step1_df.count()} patients with tafamidis >= 2019-06-01")
```
