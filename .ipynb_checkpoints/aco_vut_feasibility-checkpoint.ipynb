```python
tafamidis_codes = [str(code) for code in tafamidis_codes]
```

```python
# STEP 1: Get first tafamidis code on/after 2019-06-01
step1_df = (
    session.table("DATA_202511.MEDICATION")
    .filter(col("CODE_SYSTEM") == "RxNorm")
    .filter(col("CODE").isin(tafamidis_codes))
    .filter(col("START_DATE") >= "2019-06-01")
    .filter(col("START_DATE").is_not_null())
    .group_by("PATIENT_ID")
    .agg(s_min("START_DATE").alias("INDEX_DATE"))
)
print(f"Step 1: {step1_df.count()} patients with tafamidis >= 2019-06-01")
```

```python
session.table("DATA_202511.MEDICATION").filter(col("CODE_SYSTEM") == "RxNorm").filter(col("CODE").isin(tafamidis_codes)).filter(col("START_DATE") >= "2019-06-01").filter(col("START_DATE").is_not_null()).select('CODE').distinct().show()
```

## Step 2


```python
# STEP 2: Of Step 1, exclude those with tafamidis BEFORE INDEX_DATE
prior_tafs = (
    step1_df
    .join(
        session.table("DATA_202511.MEDICATION")
        .filter(col("CODE_SYSTEM") == "RxNorm")
        .filter(col("CODE").isin(tafamidis_codes)),
        on="PATIENT_ID"
    )
    .filter(col("START_DATE") < col("INDEX_DATE"))
    .select("PATIENT_ID").distinct()
)

step2_df = step1_df.join(prior_tafs, on="PATIENT_ID", how="anti")
print(f"Step 2: {step2_df.count()} patients after excluding prior tafamidis")
```

## Step 3

```python
eplontersen_codes = list(pd.read_excel("Aco_Vutri_TriNetX_Feas._codelists_JUL2026.xlsx", sheet_name = "Eplontersen")["Concept Code"])
eplontersen_codes = [str(code) for code in eplontersen_codes ]
```

```python
inotersen_codes = list(pd.read_excel("Aco_Vutri_TriNetX_Feas._codelists_JUL2026.xlsx", sheet_name = "Inotersen")["Concept Code"])
inotersen_codes = [str(code) for code in inotersen_codes]
```

```python
patisiran_codes = list(pd.read_excel("Aco_Vutri_TriNetX_Feas._codelists_JUL2026.xlsx", sheet_name = "Patisiran")["Concept Code"])
patisiran_codes = [str(code) for code in patisiran_codes]
```

```python
other_drugs = eplontersen_codes + inotersen_codes + patisiran_codes
```

```python
exclusions = (
    step2_df
    .join(
        session.table("DATA_202511.MEDICATION")
        .filter(col("CODE_SYSTEM") == "RxNorm")
        .filter(col("CODE").isin(other_drugs)),
        on="PATIENT_ID"
    )
    .filter(col("START_DATE") <= col("INDEX_DATE"))
    .select("PATIENT_ID").distinct()
)

step3_df = step2_df.join(exclusions, on="PATIENT_ID", how="anti")
print(f"Step 3: {step3_df.count()} after Step 3")
```

## Step 4

```python
# ICD-10-CM codes
icd10_codes = [
    "I50", "I50.1", "I50.2", "I50.20", "I50.21", "I50.22", "I50.23",
    "I50.3", "I50.30", "I50.31", "I50.32", "I50.33",
    "I50.4", "I50.40", "I50.41", "I50.42", "I50.43",
    "I50.8", "I50.81", "I50.810", "I50.811", "I50.812", "I50.813", "I50.814",
    "I50.82", "I50.83", "I50.84", "I50.89", "I50.9",
    "I11.0", "I13.0", "I13.2",
]

# ICD-9-CM codes
icd9_codes = [
    "428", "428.0", "428.1", "428.2", "428.20", "428.21", "428.22", "428.23",
    "428.3", "428.30", "428.31", "428.32", "428.33",
    "428.4", "428.40", "428.41", "428.42", "428.43",
    "428.9",
    "402.01", "402.11", "402.91",
    "404.01", "404.03", "404.11", "404.13", "404.91", "404.93",
]
```

```python
# Build lookup table: Code | Code_System
icd10_df = pd.DataFrame({"Code": icd10_codes, "Code_System": "ICD-10-CM"})
icd9_df = pd.DataFrame({"Code": icd9_codes, "Code_System": "ICD-9-CM"})

hf_codes_lookup = pd.concat([icd10_df, icd9_df], ignore_index=True)

# Extract code lists by system for use in Snowpark filtering
hf_codes_icd10 = hf_codes_lookup[hf_codes_lookup["Code_System"] == "ICD-10-CM"]["Code"].tolist()
hf_codes_icd9  = hf_codes_lookup[hf_codes_lookup["Code_System"] == "ICD-9-CM"]["Code"].tolist()
```

```python
# --- Code lists ---
# Amyloidosis codes (note: differs slightly between step 4 and 4a)
amyloidosis_codes_step4  = ["E85", "E85.0", "E85.2", "E85.4", "E85.82", "E85.89", "E85.9"]
amyloidosis_codes_step4a = ["E85", "E85.0", "E85.2", "E85.4", "E85.82", "E85.9"]  # no E85.89?
```

```python
# Cardiomyopathy/HF codes = your HF_codes list (from Excel) + these extra I-codes
extra_cm_codes = ["I42.2", "I42.5", "I42.8", "I42.9", "I43"]
```

```python
cm_codes_icd10 = hf_codes_icd10 + extra_cm_codes   
cm_codes_icd9  = hf_codes_icd9                   
```

```python
# --- Pull diagnosis records prior to/on index ---

def get_amyloidosis_records(codes):
    return (
        step3_df
        .join(
            session.table("DATA_202511.DIAGNOSIS")
            .filter(col("CODE").isin(codes))
            .filter(col("DATE").is_not_null()),
            on="PATIENT_ID"
        )
        .filter(col("DATE") <= col("INDEX_DATE"))
        .select("PATIENT_ID", col("CODE").alias("E85_CODE"), col("DATE").alias("E85_DATE"))
    )
```

```python
def get_cm_records():
    return (
        step3_df
        .join(
            session.table("DATA_202511.DIAGNOSIS")
            .filter(
                (col("CODE_SYSTEM") == "ICD-10-CM") & col("CODE").isin(cm_codes_icd10)
                | (col("CODE_SYSTEM") == "ICD-9-CM") & col("CODE").isin(cm_codes_icd9)
            )
            .filter(col("DATE").is_not_null()),
            on="PATIENT_ID"
        )
        .filter(col("DATE") <= col("INDEX_DATE"))
        .select("PATIENT_ID", col("CODE").alias("CM_CODE"), col("CODE_SYSTEM").alias("CM_CODE_SYSTEM"), col("DATE").alias("CM_DATE"))
    )
```

```python
cm_records = get_cm_records() #same for 4 and 4a
```

```python
# ============ STEP 4: any time apart ============
e85_records_4 = get_amyloidosis_records(amyloidosis_codes_step4)

step4_df = e85_records_4.join(cm_records, on="PATIENT_ID", how="inner")

print(f"Step 4: {step4_df.select('PATIENT_ID').distinct().count()} patients")
print("\nBreakdown by E85 code + CM code:")
step4_df.group_by("E85_CODE", "CM_CODE", "CM_CODE_SYSTEM").count().order_by(col("COUNT").desc()).show(30)
```

```python
print("Step 4 — Most common E85 codes:")
step4_df.select("PATIENT_ID", "E85_CODE").distinct() \
    .group_by("E85_CODE").count() \
    .order_by(col("COUNT").desc()).show(20)
```

```python
print("Step 4 — Most common CM codes:")
step4_df.select("PATIENT_ID", "CM_CODE", "CM_CODE_SYSTEM").distinct() \
    .group_by("CM_CODE", "CM_CODE_SYSTEM").count() \
    .order_by(col("COUNT").desc()).show(30)
```

### Amyloidosis analysis

```python
# Population: patients meeting Step 4 (swap for step4a_df as needed)
population_ids = step4_df.select("PATIENT_ID").distinct()
total_n = population_ids.count()
print(f"Total patients in population: {total_n}")

# Distinct (PATIENT_ID, E85_CODE) pairs for this population
patient_e85_codes = (
    e85_records_4
    .join(population_ids, on="PATIENT_ID", how="inner")
    .select("PATIENT_ID", "E85_CODE")
    .distinct()
)

# How many distinct amyloidosis codes does each patient have?
patient_code_counts = (
    patient_e85_codes
    .group_by("PATIENT_ID")
    .agg(count_distinct("E85_CODE").alias("N_DISTINCT_CODES"))
)
```

```python
# ---- TABLE 1: ≥1 code (any) ----
table_any = (
    patient_e85_codes
    .group_by("E85_CODE")
    .agg(count_distinct("PATIENT_ID").alias("N_ANY"))
    .order_by(col("N_ANY").desc())
)
print("Table 1: >=1 code (any)")
table_any.show(20)
```

```python
# ---- TABLE 2: only this code (patient has exactly 1 distinct code, and it's this one) ----
only_code_patients = (
    patient_e85_codes
    .join(patient_code_counts, on="PATIENT_ID", how="inner")
    .filter(col("N_DISTINCT_CODES") == 1)
)

table_only = (
    only_code_patients
    .group_by("E85_CODE")
    .agg(count_distinct("PATIENT_ID").alias("N_ONLY"))
    .order_by(col("N_ONLY").desc())
)
print("Table 2: only this code")
table_only.show(20)
```

### Cardiomyopathy analysis

```python
# Distinct (PATIENT_ID, CM_CODE) pairs for this population
# Using cm_records (already filtered to DATE <= INDEX_DATE)
patient_cm_codes = (
    cm_records
    .join(population_ids, on="PATIENT_ID", how="inner")
    .select("PATIENT_ID", "CM_CODE")
    .distinct()
)

# How many distinct CM codes does each patient have (out of the full CM list)?
patient_cm_code_counts = (
    patient_cm_codes
    .group_by("PATIENT_ID")
    .agg(count_distinct("CM_CODE").alias("N_DISTINCT_CM_CODES"))
)
```

```python
# ---- TABLE 1: >=1 code (any), restricted to the I42/I43 cardiomyopathy codes ----
cm_codes_of_interest = ["I42.2", "I42.5", "I42.8", "I42.9", "I43"]

table_cm_any = (
    patient_cm_codes
    .filter(col("CM_CODE").isin(cm_codes_of_interest))
    .group_by("CM_CODE")
    .agg(count_distinct("PATIENT_ID").alias("N_ANY"))
    .order_by(col("N_ANY").desc())
)
print("Table 1: >=1 cardiomyopathy code (any)")
table_cm_any.show(20)
```

```python
# ---- TABLE 2: only this code (patient's ONLY CM code overall is this one) ----
only_cm_code_patients = (
    patient_cm_codes
    .join(patient_cm_code_counts, on="PATIENT_ID", how="inner")
    .filter(col("N_DISTINCT_CM_CODES") == 1)
)

table_cm_only = (
    only_cm_code_patients
    .filter(col("CM_CODE").isin(cm_codes_of_interest))
    .group_by("CM_CODE")
    .agg(count_distinct("PATIENT_ID").alias("N_ONLY"))
    .order_by(col("N_ONLY").desc())
)
print("Table 2: only this cardiomyopathy code")
table_cm_only.show(20)
```

## Step 4a

```python
# ============ STEP 4a: within 2 years ============
e85_records_4a = get_amyloidosis_records(amyloidosis_codes_step4a)

step4a_pairs = e85_records_4a.join(cm_records, on="PATIENT_ID", how="inner")

step4a_df = step4a_pairs.filter(
    s_abs(datediff("YEAR", col("E85_DATE"), col("CM_DATE"))) <= 2
)
```

```python
print(f"\nStep 4a: {step4a_df.select('PATIENT_ID').distinct().count()} patients")
print("\nBreakdown by E85 code + CM code (within 2 years):")
step4a_df.group_by("E85_CODE", "CM_CODE", "CM_CODE_SYSTEM").count().order_by(col("COUNT").desc()).show(30)
```

# hATTR-PN

## Step 1

```python
vutrisiran_codes = list(pd.read_excel("Aco_Vutri_TriNetX_Feas._codelists_JUL2026.xlsx", sheet_name = "Vutrisiran")["Concept Code"])
vutrisiran_codes = [str(code) for code in vutrisiran_codes]
```

```python
# STEP 1: First vutrisiran code BEFORE 2025-03-01
step1_df = (
    session.table("DATA_202511.MEDICATION")
    .filter(col("CODE_SYSTEM") == "RxNorm")
    .filter(col("CODE").isin(vutrisiran_codes))
    .filter(col("START_DATE") < "2025-03-01")
    .filter(col("START_DATE").is_not_null())
    .group_by("PATIENT_ID")
    .agg(s_min("START_DATE").alias("INDEX_DATE"))
)
print(f"Step 1: {step1_df.count()} patients")
```

```python
session.table("DATA_202511.MEDICATION").filter(col("CODE_SYSTEM") == "RxNorm").filter(col("CODE").isin(vutrisiran_codes)).filter(col("START_DATE") >= "2019-06-01").filter(col("START_DATE").is_not_null()).select('CODE').distinct().show()
```

## Step 2

```python
# STEP 2: Exclude patients with vutrisiran BEFORE their index date
prior_vutrisiran = (
    step1_df
    .join(
        session.table("DATA_202511.MEDICATION")
        .filter(col("CODE_SYSTEM") == "RxNorm")
        .filter(col("CODE").isin(vutrisiran_codes)),
        on="PATIENT_ID"
    )
    .filter(col("START_DATE") < col("INDEX_DATE"))
    .select("PATIENT_ID").distinct()
)

step2_df = step1_df.join(prior_vutrisiran, on="PATIENT_ID", how="anti")
print(f"Step 2: {step2_df.count()} patients after excluding prior vutrisiran")
```

## Step 3

```python
e85_1_code = ["E85.1"]
e85_group_codes = ["E85", "E85.2", "E85.89", "E85.9"]
neuro_codes = [
    "G60", "G60.3", "G60.9",
    "G61", "G61.8", "G61.82", "G61.89", "G61.9",
    "G62", "G62.8", "G62.89", "G62.9",
    "G63", "G64"
]
```

```python
diagnosis_tbl = session.table("DATA_202511.DIAGNOSIS")
```

```python
# ---- E85.1 records (standalone qualifier) ----
e85_1_records = (
    step2_df
    .join(
        diagnosis_tbl.filter(col("CODE").isin(e85_1_code)).filter(col("DATE").is_not_null()),
        on="PATIENT_ID"
    )
    .filter(col("DATE") <= col("INDEX_DATE"))
    .select("PATIENT_ID", col("CODE").alias("E85_1_CODE"), col("DATE").alias("E85_1_DATE"))
)

# ---- E85 group records (for the AND clause) ----
e85_group_records = (
    step2_df
    .join(
        diagnosis_tbl.filter(col("CODE").isin(e85_group_codes)).filter(col("DATE").is_not_null()),
        on="PATIENT_ID"
    )
    .filter(col("DATE") <= col("INDEX_DATE"))
    .select("PATIENT_ID", col("CODE").alias("E85_CODE"), col("DATE").alias("E85_DATE"))
)

# ---- Neuropathy (G6x) records ----
neuro_records = (
    step2_df
    .join(
        diagnosis_tbl.filter(col("CODE").isin(neuro_codes)).filter(col("DATE").is_not_null()),
        on="PATIENT_ID"
    )
    .filter(col("DATE") <= col("INDEX_DATE"))
    .select("PATIENT_ID", col("CODE").alias("NEURO_CODE"), col("DATE").alias("NEURO_DATE"))
)
```

```python
# Patients meeting E85.1 criterion
pathway_a_patients = e85_1_records.select("PATIENT_ID").distinct()

# Patients meeting E85 group AND neuro group (any time apart)
e85_group_pairs = e85_group_records.join(neuro_records, on="PATIENT_ID", how="inner")
pathway_b_patients = e85_group_pairs.select("PATIENT_ID").distinct()

# Union of both pathways (patients meeting EITHER condition)
step3_patient_ids = pathway_a_patients.union(pathway_b_patients).distinct()

step3_df = step2_df.join(step3_patient_ids, on="PATIENT_ID", how="inner")
print(f"Step 3: {step3_df.count()} patients")
```

### Amyloidosis

```python
# Population: patients meeting Step 3 (swap for step3a_df as needed)
population_ids = step3_df.select("PATIENT_ID").distinct()

# Full amyloidosis code list for this table (E85.1 + E85 group)
all_amyloidosis_codes = ["E85.1", "E85", "E85.2", "E85.89", "E85.9"]
```

```python
# AMYLOIDOSIS TABLE

# Distinct (PATIENT_ID, CODE) pairs for amyloidosis codes, restricted to population, <= index date
patient_amyloid_codes = (
    diagnosis_tbl
    .filter(col("CODE").isin(all_amyloidosis_codes))
    .filter(col("DATE").is_not_null())
    .join(step3_df.select("PATIENT_ID", "INDEX_DATE"), on="PATIENT_ID", how="inner")
    .filter(col("DATE") <= col("INDEX_DATE"))
    .select("PATIENT_ID", col("CODE").alias("AMYLOID_CODE"))
    .distinct()
)

# How many distinct amyloidosis codes does each patient have?
patient_amyloid_code_counts = (
    patient_amyloid_codes
    .group_by("PATIENT_ID")
    .agg(count_distinct("AMYLOID_CODE").alias("N_DISTINCT_CODES"))
)
```

```python
# ---- TABLE 1: >=1 code (any) ----
table_amyloid_any = (
    patient_amyloid_codes
    .group_by("AMYLOID_CODE")
    .agg(count_distinct("PATIENT_ID").alias("N_ANY"))
    .order_by(col("N_ANY").desc())
)
print("Amyloidosis Table 1: >=1 code (any)")
table_amyloid_any.show(20)
```

```python
# ---- TABLE 2: only this code ----
only_amyloid_patients = (
    patient_amyloid_codes
    .join(patient_amyloid_code_counts, on="PATIENT_ID", how="inner")
    .filter(col("N_DISTINCT_CODES") == 1)
)

table_amyloid_only = (
    only_amyloid_patients
    .group_by("AMYLOID_CODE")
    .agg(count_distinct("PATIENT_ID").alias("N_ONLY"))
    .order_by(col("N_ONLY").desc())
)
print("Amyloidosis Table 2: only this code")
table_amyloid_only.show(20)
```

### Neuropathy

```python
patient_neuro_codes = (
    diagnosis_tbl
    .filter(col("CODE").isin(neuro_codes))
    .filter(col("DATE").is_not_null())
    .join(step3_df.select("PATIENT_ID", "INDEX_DATE"), on="PATIENT_ID", how="inner")
    .filter(col("DATE") <= col("INDEX_DATE"))
    .select("PATIENT_ID", col("CODE").alias("NEURO_CODE"))
    .distinct()
)

patient_neuro_code_counts = (
    patient_neuro_codes
    .group_by("PATIENT_ID")
    .agg(count_distinct("NEURO_CODE").alias("N_DISTINCT_CODES"))
)
```

```python
# ---- TABLE 1: >=1 code (any) ----
table_neuro_any = (
    patient_neuro_codes
    .group_by("NEURO_CODE")
    .agg(count_distinct("PATIENT_ID").alias("N_ANY"))
    .order_by(col("N_ANY").desc())
)
print("Neuropathy Table 1: >=1 code (any)")
table_neuro_any.show(20)
```

```python
# ---- TABLE 2: only this code ----
only_neuro_patients = (
    patient_neuro_codes
    .join(patient_neuro_code_counts, on="PATIENT_ID", how="inner")
    .filter(col("N_DISTINCT_CODES") == 1)
)

table_neuro_only = (
    only_neuro_patients
    .group_by("NEURO_CODE")
    .agg(count_distinct("PATIENT_ID").alias("N_ONLY"))
    .order_by(col("N_ONLY").desc())
)
print("Neuropathy Table 2: only this code")
table_neuro_only.show(20)
```

## Step 3a

```python
# For pathway B, get first date of each per patient, then check 2-year window
e85_group_first = (
    e85_group_records
    .group_by("PATIENT_ID")
    .agg(s_min("E85_DATE").alias("E85_DATE_FIRST"))
)

neuro_first = (
    neuro_records
    .group_by("PATIENT_ID")
    .agg(s_min("NEURO_DATE").alias("NEURO_DATE_FIRST"))
)

pathway_b_within_2yr = (
    e85_group_first
    .join(neuro_first, on="PATIENT_ID", how="inner")
    .filter(s_abs(datediff("YEAR", col("E85_DATE_FIRST"), col("NEURO_DATE_FIRST"))) <= 2)
    .select("PATIENT_ID")
)
```

```python
# Union: E85.1 pathway (no time restriction) OR pathway B within 2 years
step3a_patient_ids = pathway_a_patients.union(pathway_b_within_2yr).distinct()

step3a_df = step2_df.join(step3a_patient_ids, on="PATIENT_ID", how="inner")
print(f"Step 3a: {step3a_df.count()} patients")
```
